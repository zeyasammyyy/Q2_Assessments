```python
roblox_games = ["Tower of Hell", "Murder Mystery", "Flee the Facility", "Pet Store Tycoon", "Adopt Me"]

print("=== Method 1: Looping by Game ===")
for game in roblox_games:
  print(f"Game: {game}")

print("\n=== Method 2: Looping by Index (with Position Labels) ===")
for i in range(len(roblox_games)):
  print(f"Index {i} -> {roblox_games[i]}")
```

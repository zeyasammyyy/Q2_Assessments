```python
# Problem 1: Personal Savings Tracker

def update_savings(current_total, deposit):
    return current_total + deposit

def check_goal(total, target):
    if total >= target:
        return "Goal reached!"
    else:
        return "Keep saving!"

target = float(input("Enter savings target: ₱"))
total_savings = 0.0

print()
while True:
    deposit = float(input("Enter deposit (0 to finish): ₱"))
    if deposit == 0:
        break
    total_savings = update_savings(total_savings, deposit)
    print(f"Updated savings: ₱{total_savings:.2f}\n")

print(f"\nTotal savings: ₱{total_savings:.2f}")

if total_savings < target:
    remaining = target - total_savings
    print(f"Remaining amount needed: ₱{remaining:.2f}")

print(check_goal(total_savings, target))

# Problem 2: Classroom Energy Usage Analyzer

def calculate_energy(power_watts, hours_used):
    return (power_watts / 1000) * hours_used

def classify_usage(energy):
    if energy < 0.5:
        return "Low"
    elif energy <= 2.0:
        return "Moderate"
    else:
        return "High"

num_devices = int(input("How many devices? "))
total_energy = 0.0
highest_energy = -1.0
highest_device = ""

print()
for i in range(1, num_devices + 1):
    name = input(f"Device {i} name: ")
    power = float(input(f"Power rating (watts): "))
    hours = float(input(f"Hours used: "))

    energy = calculate_energy(power, hours)
    classification = classify_usage(energy)

    print(f"Energy consumption: {energy:.2f} kWh")
    print(f"Classification: {classification}\n")

    total_energy += energy

    if energy > highest_energy:
        highest_energy = energy
        highest_device = name

print(f"Total energy consumption: {total_energy:.2f} kWh")
print(f"Highest consumption device: {highest_device} ({highest_energy:.2f} kWh)")

# Problem 3: Water Refilling Station Sales Tracker

def calculate_sale(quantity, price_per_container):
    return quantity * price_per_container

def calculate_discount(total):
    if total >= 500:
        return total * 0.05
    return 0.0

price_per_container = float(input("Price per container: ₱"))
total_sales_before = 0.0
total_discounts = 0.0

print()
while True:
    qty = int(input("Enter quantity sold (0 to finish): "))
    if qty == 0:
        break

    sale = calculate_sale(qty, price_per_container)
    discount = calculate_discount(sale)
    amount_due = sale - discount

    print(f"Sale: ₱{sale:.2f}")
    print(f"Discount: ₱{discount:.2f}")
    print(f"Amount due: ₱{amount_due:.2f}\n")

    total_sales_before += sale
    total_discounts += discount

net_sales = total_sales_before - total_discounts

print(f"\nTotal sales before discounts: ₱{total_sales_before:.2f}")
print(f"Total discounts: ₱{total_discounts:.2f}")
print(f"Net sales: ₱{net_sales:.2f}")

# Problem 4: Reading Time and Progress Analyzer

def calculate_reading_speed(pages, minutes):
    if minutes == 0:
        return 0.0
    return pages / minutes

def classify_speed(speed):
    if speed < 1:
        return "Developing"
    elif speed <= 2:
        return "Steady"
    else:
        return "Fast"

num_sessions = int(input("How many reading sessions? "))
total_pages = 0
total_minutes = 0

print()
for i in range(1, num_sessions + 1):
    pages = int(input(f"Session {i} pages read: "))
    minutes = int(input(f"Minutes spent: "))

    speed = calculate_reading_speed(pages, minutes)
    classification = classify_speed(speed)

    print(f"Reading speed: {speed:.2f} pages/minute")
    print(f"Classification: {classification}\n")

    total_pages += pages
    total_minutes += minutes

overall_speed = calculate_reading_speed(total_pages, total_minutes)

print(f"Total pages read: {total_pages}")
print(f"Total reading time: {total_minutes} minutes")
print(f"Overall reading speed: {overall_speed:.2f} pages/minute")

# Problem 5: School Event Attendance Monitor

def calculate_attendance_rate(attendees, expected):
    if expected == 0:
        return 0.0
    return (attendees / expected) * 100

def classify_attendance(rate):
    if rate < 60:
        return "Low"
    elif rate < 90:
        return "Moderate"
    else:
        return "High"

num_sessions = int(input("How many sessions? "))
total_attendees = 0
total_expected = 0
high_count = 0

print()
for i in range(1, num_sessions + 1):
    expected = int(input(f"Session {i} expected attendance: "))
    attendees = int(input(f"Actual attendees: "))

    rate = calculate_attendance_rate(attendees, expected)
    classification = classify_attendance(rate)

    print(f"Attendance rate: {rate:.2f}%")
    print(f"Classification: {classification}\n")

    total_attendees += attendees
    total_expected += expected

    if classification == "High":
        high_count += 1

overall_rate = calculate_attendance_rate(total_attendees, total_expected)

print(f"Total attendees: {total_attendees}")
print(f"Total expected attendance: {total_expected}")
print(f"Overall attendance rate: {overall_rate:.2f}%")
print(f"High-attendance sessions: {high_count}")
```
